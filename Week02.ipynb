### 1. Tampilkan citra hanya pada kombinasi kanal Merah-Biru dan kanal Hijau-Biru saja, dengan kanal yang tidak dipakai dinolkan. Susun keduanya berdampingan dalam satu figure yang diberi judul.

```python
b,g,r = cv.split(image_citra)

# array kosong
zeros=np.zeros_like(b)

# merah biru
img_red_blue=cv.merge([r,zeros,b])
# ijo biru
img_green_blue=cv.merge([zeros,g,b])

fig,axes=plt.subplots(1,2,figsize=(15,7))
axes[0].imshow(img_red_blue)
axes[0].set_title('Merah-Biru')
axes[1].imshow(img_green_blue)
axes[1].set_title('Hijau-Biru')
plt.tight_layout()
plt.show()
```

### 2. Tampilkan citra dalam tiga bentuk pencerminan: vertikal, horizontal, dan keduanya, dalam satu figure berisi tiga subplot yang masing-masing diberi label.

```python
img_citra_rgb=cv.cvtColor(image_citra,cv.COLOR_BGR2RGB)
img_flip_v=cv.flip(img_citra_rgb,0)
img_flip_h=cv.flip(img_citra_rgb,1)
img_flip_vh=cv.flip(img_citra_rgb,-1)

fig,axes=plt.subplots(1,3,figsize=(15,5))
axes[0].imshow(img_flip_v)
axes[0].set_title('Flip Vertikal')
axes[1].imshow(img_flip_h)
axes[1].set_title('Flip Horizontal')
axes[2].imshow(img_flip_vh)
axes[2].set_title('Flip Vertikal dan Horizontal')
plt.tight_layout()
plt
```
