Pertanyaan:
Apakah proses pengecilan resolusi lalu pembesaran kembali mengubah nilai rata-rata intensitas piksel serta menghilangkan variasi detail pada grayscale?

Rancangan Program:
1. Baca file citra KTM dalam mode grayscale menggunakan OpenCV.
2. Hitung nilai rata-rata intensitas piksel dan standar deviasi pada citra asli menggunakan NumPy.
3. Perkecil dimensi citra menjadi 25 persen dari ukuran asli memakai cv2.resize dengan interpolasi INTER_AREA.
4. Perbesar kembali dimensi citra ke ukuran semula memakai cv2.resize dengan interpolasi INTER_CUBIC.
5. Hitung nilai rata-rata intensitas, standar deviasi, dan Mean Squared Error (MSE) antara citra asli dan citra rekonstruksi.
6. Tampilkan perbandingan angka hasil perhitungan.

Data hasilnya:

```python
img_gray=cv.imread('ktm.jpeg',cv.IMREAD_GRAYSCALE)
h,w=img_gray.shape[:2]
img_small=cv.resize(img_gray,(int(w*0.25),int(h*0.25)),interpolation=cv.INTER_AREA)
img_restored=cv.resize(img_small,(w,h),interpolation=cv.INTER_CUBIC)

mean_asli=np.mean(img_gray)
std_asli=np.std(img_gray)

mean_restored=np.mean(img_restored)
std_restored=np.std(img_restored)

mse=np.mean((img_gray-img_restored)**2)
print('Nilai rata-rata intensitas asli: ',mean_asli)
print('Nilai rata-rata intensitas restored: ',mean_restored)
print('Nilai standar deviasi asli: ',std_asli)
print('Nilai standar deviasi restored: ',std_restored)
print('Nilai MSE: ',mse)
```

Kesimpulan:
Nilai rata-rata intensitas hanya bergeser sekitar 0.001 dari 173.359 menjadi 173.360 saat proses pengecilan lalu pembesaran citra. pada saat proses interpolasi, penurunan variasi nilai standar deviasi dari 74.240 menjadi 73.317. Nilai MSE sebesar 15.498 membuktikan bahwa interpolasi pembesaran tidak mampu mengembalikan nilai piksel asli secara sempurna.


Studi Kasus 1: Panitia sebuah kegiatan perlu membubuhkan label identitas pada
ratusan foto dokumentasi. Foto berasal dari beberapa kamera dan ponsel,
sehingga ukurannya bermacam-macam. Label harus terlihat seimbang pada
semua foto tanpa disetel satu per satu. Fungsi yang sama dikenakan pada dua
citra berbeda ukuran. Tinggi bar dan ukuran huruf ikut menyesuaikan, sehingga
proporsinya tetap sama. Silakan melakukan akuisisi data sendiri, setiap anak
harus menggunakan image yang berbeda-beda

```python
img1=cv.imread('alam.jpeg')
img2=cv.imread('kucing.jpeg')

hasil1=cv.cvtColor(img1,cv.COLOR_BGR2RGB)
hasil2=cv.cvtColor(img2,cv.COLOR_BGR2RGB)

def label_dokumentasi(images, teks="Dokumentasi PCVK - 2026"):
  h,w=images.shape[:2]
  hasil=images.copy()

  tinggi_bar=h//12
  cv.rectangle(hasil, (0,h-tinggi_bar), (w,h), (0,0,0), -1)

  skala_font=tinggi_bar/35.0
  tebal_font=max(1, int(tinggi_bar/25))

  (lebar_teks, tinggi_teks), _ = cv.getTextSize(teks, cv.FONT_HERSHEY_SIMPLEX, skala_font, tebal_font)

  lebar_maksimal = int(w * 0.9)
  if lebar_teks > lebar_maksimal:
    skala_font = skala_font * (lebar_maksimal / lebar_teks)
    (lebar_teks, tinggi_teks), _ = cv.getTextSize(teks, cv.FONT_HERSHEY_SIMPLEX, skala_font, tebal_font)

  x=int(w*0.02)
  y=h-int(tinggi_bar*0.35)

  cv.putText(hasil, teks, (x,y), cv.FONT_HERSHEY_SIMPLEX, skala_font, (255,255,255), tebal_font, cv.LINE_AA)
  return hasil

hasil1_fix=label_dokumentasi(hasil1)
hasil2_fix=label_dokumentasi(hasil2)
```

Studi Kasus 2: Marketplace seperti Tokopedia dan Shopee menampilkan foto
produk dalam bingkai persegi 1:1. Foto penjual yang berukuran potrait atau
lanskap akan terpotong atau terlihat tidak proporsional bila diunggah begitu saja.
Bangun program yang mengubah foto berukuran apa pun menjadi thumbnail
persegi tanpa mengubah rasio aslinya, dengan sisa ruang diisi kanvas berwarna.
Dalam hal ini, rasio asli tetap terjaga, citra terpasang tepat di tengah kanvas,
tambahkan logo nama toko berukuran proporsional pada gambar.

```python
h,w=img2.shape[:2]
sisi=max(h,w)
kanvas=np.full((sisi,sisi,3), (255,255,255), dtype=np.uint8)
sisi_x=int((sisi-w)/2)
sisi_y=int((sisi-h)/2)
kanvas[sisi_y:sisi_y+h,sisi_x:sisi_x+w]=img2

skala_font=sisi/850
tebal_font=5

(lebar_teks, tinggi_teks), _ = cv.getTextSize("Tokopedia", cv.FONT_HERSHEY_SIMPLEX, skala_font, tebal_font)

jarak_tepi=int(sisi*0.035)
radius_logo=int(tinggi_teks*0.75)

logo_x=jarak_tepi+radius_logo
logo_y=jarak_tepi+radius_logo

cv.circle(kanvas, (logo_x,logo_y), radius_logo, (0,255,0), -1, lineType=cv.LINE_AA)

teks_x=logo_x+radius_logo+int(sisi*0.02)
teks_y=logo_y+radius_logo+(tinggi_teks//2)

cv.putText(kanvas, "Toko", (teks_x,teks_y), cv.FONT_HERSHEY_SIMPLEX, skala_font, (125,125,125), tebal_font, cv.LINE_AA)

plt.imshow(kanvas)
```
